```python
import pandas as pd
import geopandas as gpd
import osmnx as ox
import matplotlib.pyplot as plt

from shapely.geometry import Point
```

```python
df = pd.read_csv(r"C:\Users\kashv\Desktop\Delhi_UHI_Project\Data\Raw\Delhi_UHI_Final_Dataset.csv")

df.head()
```

```python
geometry = [
    Point(xy)
    for xy in zip(df["Longitude"], df["Latitude"])
]

gdf = gpd.GeoDataFrame(
    df,
    geometry=geometry,
    crs="EPSG:4326"
)

gdf.head()
```

```python
delhi = ox.geocode_to_gdf("Delhi, India")

roads = ox.features_from_place(
    "Delhi, India",
    tags={"highway": True}
)

roads = roads[
    roads.geometry.type.isin(
        ["LineString", "MultiLineString"]
    )
]
```

```python
roads.head()
```

```python
roads.plot(figsize=(10,10))
```

```python
# Convert to metric CRS (UTM Zone 43N)
gdf = gdf.to_crs(epsg=32643)
roads = roads.to_crs(epsg=32643)
```

```python
road_union = roads.union_all()
gdf["buffer"] = gdf.geometry.buffer(500)
```

```python
road_density = []

for buffer in gdf["buffer"]:
    clipped = roads.clip(buffer)
    total_length = clipped.length.sum() # meters
    area = buffer.area                  # square meters
    density = total_length / area
    road_density.append(density)

gdf["RoadDensity"] = road_density
```

```python
gdf["RoadDensity"].describe()
```

```python
plt.hist(gdf["RoadDensity"], bins=30, edgecolor="black")

plt.title("Road Density Distribution")
plt.xlabel("Road Density")
plt.ylabel("Number of Samples")

plt.show()
```

```python
gdf.drop(columns="buffer", inplace=True)
```

```python
gdf.head()
```

```python
place = "Delhi, India"
tags = {"building": True}
buildings = ox.features_from_place(place, tags)

# Keep only polygon buidings
buildings = buildings[
    buildings.geometry.type.isin(["Polygon", "MultiPolygon"])
]
```

```python
buildings.head()
```

```python
buildings.plot(figsize=(8,8))
```

```python
buildings = buildings.to_crs(epsg=32643)
gdf = gdf.to_crs(epsg=32643)

building_sindex = buildings.sindex
radius = 250 # meters

building_density = []

for point in gdf.geometry:
    buffer = point.buffer(radius)

    possible_matches = list(building_sindex.intersection(buffer.bounds))
    nearby_buildings = buildings.iloc[possible_matches]

    nearby_buildings = nearby_buildings[
        nearby_buildings.intersects(buffer)
    ]

    total_area = nearby_buildings.intersection(buffer).area.sum()

    density = total_area / buffer.area

    building_density.append(density)
gdf["BuildingDensity"] = building_density
```

```python
gdf["BuildingDensity"].describe()
```

```python
plt.hist(gdf["BuildingDensity"], bins=30, edgecolor="black")
plt.xlabel("Building Density")
plt.ylabel("Frequency")
plt.title("Building Density Distribution")
plt.show()
```

```python
gdf.sample(10)
```

```python
gdf.drop("geometry",axis=1, inplace=True)
```

```python
lulc_map = {
    0 : "Water",
    1 : "Trees",
    2 : "Grass",
    3 : "Flooded Vegetation",
    4 : "Crops",
    5 : "Shrub",
    6 : "Built Area",
    7 : "Bare Ground",
    8 : "Snow"
}

gdf["LULC_Name"] = gdf["LULC"].map(lulc_map)
```
